# Parte 5 — Definir os parâmetros da consulta

Esta parte define os parâmetros utilizados para consultar os resultados de uma medição pública da API RIPE Atlas.

- `MEASUREMENT_ID`: identificador numérico da medição pública;
- `HORAS_COLETA`: período anterior ao momento da execução;
- `FORMATO_TABELAR`: formato escolhido para salvar os dados;
- `TIMEOUT_REQUISICAO`: tempo máximo de espera da requisição, em segundos.

O período da consulta utiliza timestamps Unix em UTC.


```python
from datetime import datetime, timezone, timedelta

# Parâmetros definidos pela equipe.
MEASUREMENT_ID = 2000000
HORAS_COLETA = 2
FORMATO_TABELAR = "csv"  # opções: "csv" ou "parquet"
TIMEOUT_REQUISICAO = 60

fim_utc = datetime.now(timezone.utc)
inicio_utc = fim_utc - timedelta(hours=HORAS_COLETA)

URL = f"https://atlas.ripe.net/api/v2/measurements/{MEASUREMENT_ID}/results/"
PARAMETROS = {
    "start": int(inicio_utc.timestamp()),
    "stop": int(fim_utc.timestamp()),
}

assert isinstance(MEASUREMENT_ID, int) and MEASUREMENT_ID > 0
assert HORAS_COLETA > 0
assert FORMATO_TABELAR in {"csv", "parquet"}
assert TIMEOUT_REQUISICAO > 0

print("URL:", URL)
print("Início (UTC):", inicio_utc.isoformat())
print("Fim (UTC):", fim_utc.isoformat())
print("Parâmetros:", PARAMETROS)
```

## Justificativa dos parâmetros

Foi escolhida a medição pública identificada pelo número `7714918`, conforme definido pela equipe. O período de coleta foi definido em duas horas anteriores ao momento da execução, permitindo obter dados recentes sem solicitar um volume excessivo de registros. O formato escolhido foi CSV por ser simples de visualizar, compartilhar e utilizar posteriormente em ferramentas de análise de dados. O tempo máximo de espera foi definido em 60 segundos, permitindo que a API tenha tempo suficiente para responder sem deixar a execução aguardando indefinidamente.
